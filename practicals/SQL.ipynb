```python
from pyspark.sql import SparkSession
spark = SparkSession.builder\
    .appName("SQL")\
    .getOrCreate()
spark.sparkContext.setLogLevel("WARN")
```

```python
columns = ['name', 'age', 'role', 'salary']
data = [
    ("Brian", 23, "AI Engineer", 32000),
    ("Saheed", 25, "Research Engineer", 40000),
    ("Stephen", 55, "Professor", 70000),
    ("Noncedo", 20, "Agronomist", 30000),
]

df =  spark.createDataFrame(data, columns)
df.show()
```

```python
df.createOrReplaceTempView("employees")
```

```python
result = spark.sql("SELECT * FROM employees")
```

```python
result.show()
```

```python
spark.sql("SELECT name FROM employees WHERE salary > 35000").show() # adding the where clause
```
