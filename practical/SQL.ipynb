Monitoring your PySpark jobs is an important part of developing a robust program. To help with this, PySpark provides some levels of logging. By default, WARN is used. The logging levels available are summarized into the following table

| Level | Description|
|-------|-------------|
| OFF | No logging at all|
|FATAL| Logs only fatal errors|
| ERROR| Shows fatal errors but also some recoverable ones|
|WARN | Adds warnings|
|INFO    | Gives runtime information
|DEBUG  | Provides debug information |
| TRACE | Will trace your jobs|
| ALL | Everything happening on your PySpark |

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, FloatType
```

```python

spark = SparkSession.builder\
    .appName("SQL")\
    .getOrCreate()
spark.sparkContext.setLogLevel("WARN")
```

```python

schema = StructType([
    StructField('name', StringType(), nullable=False),
    StructField('age', IntegerType(), True),
    StructField('role', StringType(), True),
    StructField('salary', FloatType(), True),
])


data = [
    ("Brian", 23, "AI Engineer", 32000.00),
    ("Saheed", 25, "Research Engineer", 40000.00),
    ("Stephen", 55, "Professor", 70000.00),
    ("Noncedo", 20, "Agronomist", 30000.00),
]

df =  spark.createDataFrame(data, schema=schema)
df.show()
```

```python
df.printSchema()
```

```python
df.createOrReplaceTempView("employees")
```

```python
result = spark.sql("SELECT * FROM employees")
```

```python
result.show()
```

```python
spark.sql("SELECT name FROM employees WHERE salary > 35000").show() # adding the where clause
```

```python
spark.sql("SELECT * FROM employees WHERE role LIKE '%Engineer%'").show()
```
